```python
import json

# 코퍼스 로드 (한 번만)
corpus = {}
with open("data/kbl-rag/corpus/statutes.jsonl") as f:
    for line in f:
        obj = json.loads(line)
        corpus[obj["id"]] = obj["contents"]

# 52개 쿼리별로 gold_text 일부로 검색
with open("data/kbl-rag/tasks/query_statute_matching.jsonl") as f:
    for i, line in enumerate(f):
        obj = json.loads(line)
        snippet = obj["gold_text"][20:60]  # 앞부분 헤더 제외하고 본문 일부
        hits = [doc_id for doc_id, text in corpus.items() if snippet in text]
        print(f"[{i+1:02d}] doc_id={obj['doc_id']} | {obj['gold_ref']}")
        print(f"      hits: {hits[:3]}")
```
