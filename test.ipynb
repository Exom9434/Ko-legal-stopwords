```python
from datasets import load_dataset
ds = load_dataset('lbox/kbl', 'kbl_knowledge_query_and_statute_matching_qa', trust_remote_code=True)
print('splits:', list(ds.keys()))
# 첫 레코드 구조 출력
split = list(ds.keys())[0]
item = ds[split][0]
for k, v in item.items():
    print(f'  {k}: {str(v)[:150]}')
print('총 레코드:', sum(len(ds[s]) for s in ds))
```
