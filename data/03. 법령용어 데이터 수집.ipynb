```python
import requests
import xml.etree.ElementTree as ET
import pandas as pd
import time

def fetch_legal_terms_korean_tags(user_id):
    base_url = "http://www.law.go.kr/DRF/lawSearch.do"
    all_terms = []
    page = 1
    display_count = 100  # 한 번에 요청할 최대 개수
    
    print(f"[{user_id}] ID로 데이터 수집을 시작합니다...")

    while True:
        params = {
            'OC': user_id,
            'target': 'lstrm',
            'type': 'XML',
            'display': display_count,
            'page': page,
            'dicKndCd' : '010101'
        }

        try:
            response = requests.get(base_url, params=params)
            
            if response.status_code != 200:
                print(f"페이지 {page} 요청 실패 (Code: {response.status_code})")
                break
            
            # XML 파싱
            root = ET.fromstring(response.content)
            
            # 총 개수 확인 (첫 페이지일 때만 출력)
            if page == 1:
                total_cnt = root.find('totalCnt')
                if total_cnt is not None:
                    print(f"총 수집 대상 용어 수: {total_cnt.text}개")

            # <lstrm> 태그 찾기
            items = root.findall('lstrm')
            
            if not items:
                print("더 이상 데이터가 없습니다. 수집 종료.")
                break
            
            for item in items:
                # 확인된 태그 이름들을 직접 매핑
                term_data = {
                    '법령용어ID': item.findtext('법령용어ID'),
                    '법령용어명': item.findtext('법령용어명'),
                    '법령용어상세링크': item.findtext('법령용어상세링크'),
                    '법령종류코드': item.findtext('법령종류코드'),
                    '사전구분코드': item.findtext('사전구분코드')
                }
                # 텍스트 정리 (공백 제거나 None 처리)
                for key, val in term_data.items():
                    if val:
                        term_data[key] = val.strip()
                    else:
                        term_data[key] = ""
                        
                all_terms.append(term_data)
            
            print(f"{page} 페이지 완료 (현재까지 {len(all_terms)}개 저장됨)")
            
            # 종료 조건: 이번에 가져온 개수가 요청 개수(100)보다 적으면 마지막 페이지
            if len(items) < display_count:
                break
                
            page += 1
            time.sleep(0.3)  # 너무 빠른 요청 방지

        except Exception as e:
            print(f"에러 발생: {e}")
            break

    return pd.DataFrame(all_terms)

# ==========================================
# 실행부
# ==========================================
if __name__ == "__main__":
    # [주의] 본인의 OC 아이디를 입력하세요
    MY_OC_ID = "silverbreak" 
    
    df = fetch_legal_terms_korean_tags(MY_OC_ID)
    
    if not df.empty:
        file_name = "법령용어목록_전체"
        # 엑셀 저장
        df.to_excel(f"{file_name}.xlsx", index=False)
        # CSV 저장 (한글 깨짐 방지 utf-8-sig)
        df.to_csv(f"{file_name}.csv", index=False, encoding='utf-8-sig')
        print(f"\n[완료] 총 {len(df)}개의 용어가 '{file_name}.xlsx'에 저장되었습니다.")
        
        # 데이터 샘플 출력
        print(df.head())
    else:
        print("\n수집된 데이터가 없습니다.")
```
